```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable
import numpy as np

# Schritt 1: Daten vorbereiten
# Laden Sie die neue Daten
data = pd.read_csv('Neu_Filterung_Gewichtung_Min20Data_Deutschland.csv')

# Stellen Sie sicher, dass es keine unendlichen oder NaN-Werte gibt
data['Endbewertung'].fillna(0, inplace=True)
data['Endbewertung'].replace([float('inf'), -float('inf')], 0, inplace=True)

# Schritt 2: Shapefile laden
grid = gpd.read_file('DE_Grid_ETRS89-LAEA_5km\\geogitter\\DE_Grid_ETRS89-LAEA_5km.shp')
germany = gpd.read_file('Map_Plot\\vg2500_bld.shp')

# Schritt 3: Daten zusammenführen
# Stellen Sie sicher, dass die Schlüsselspalte in beiden Datensätzen übereinstimmt
merged = grid.merge(data, left_on="id", right_on="GITTER_5km")

# Projizieren Sie die Daten, wenn nötig
merged = merged.to_crs(epsg=4326)

# Schritt 4: Karte erstellen
fig, ax = plt.subplots(1, 1, figsize=(15, 15))

germany.boundary.plot(ax=ax, color='black', linewidth=1)

# Verwenden der 'Endbewertung' Spalte für die Farbkodierung
merged.plot(column='Endbewertung', ax=ax,
            cmap='OrRd', legend=False, legend_kwds={'label': "Fahrradinfrastrukturbewertung"})

# Zeichnen Sie die Umrandung für alle Gitternetze
for _, row in merged.iterrows():
    plt.plot(*row['geometry'].exterior.xy, color='black', linewidth=0.5)

divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="5%", pad=0.1)
sm = plt.cm.ScalarMappable(cmap='OrRd', norm=plt.Normalize(vmin=min(merged['Endbewertung']), vmax=max(merged['Endbewertung'])))
sm._A = []  # Array der Datenwerte, um die Colorbar zu füllen; notwendig in manchen Matplotlib-Versionen
cbar = fig.colorbar(sm, cax=cax)

cbar.ax.tick_params(labelsize=14)
cbar.set_label('Fahrradinfrastrukturbewertung', fontsize=16)

ax.set_axis_off()

plt.show()
```
